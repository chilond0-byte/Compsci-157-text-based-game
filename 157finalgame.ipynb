```python

import os
import subprocess

audio_folder = os.path.join(os.path.dirname("/Users/nick_chilom/Downloads/audio for game"), "audio")
music_process = None

def play_scene(filename):
    global music_process
    if music_process is not None:
        music_process.terminate()
    path = os.path.join(audio_folder, filename)
    music_process = subprocess.Popen(["afplay", path])

def play_effect(filename):
    path = os.path.join(audio_folder, filename)
    subprocess.Popen(["afplay", path])


choices = []
inventory = []
knows_passage = False
knows_room = False
has_mace = False
destroyed_mace = False
caught = False

play_scene("1 .m4a")
print("THE UNIVERSITY MACE")
print("Sewanee, 1965. You tour All Saints Chapel with your father.")
print("A guide explains the University Mace.")
print("1. Ask about a hidden basement entrance")
print("2. Ask where the mace is stored")
answer = input("> ")
choices.append("Tour: " + answer)

if answer == "1":
    print("The guide tells you about a supposed hidden entrance beneath the chapel.")
    knows_passage = True
else:
    print("The guide tells you the name of the room where the mace is stored.")
    knows_room = True

print("You remember time with your father.")
print("The memory fades. You are at his procession in the chapel.")
print("The University Mace is displayed. You know its history.")
play_scene("2 final music for game.m4a")
print("After the ceremony, you can stay or go to the cemetery.")
print("1. Stay in the chapel")
print("2. Go to the cemetery")
answer = input("> ")
choices.append("After the procession: " + answer)

if answer == "2":
    play_scene("7 graveyard .m4a")
    print("At the cemetery, you can look for useful items or return.")
    print("1. Look around")
    print("2. Return to the chapel")
    answer = input("> ")
    choices.append("Cemetery: " + answer)
    if answer == "1":
        print("You find a shovel, sticks, and rocks.")
        inventory.append("shovel")
        inventory.append("sticks")
        inventory.append("rocks")
    play_scene("2 final music for game.m4a")
    print("You return to the chapel.")

print("\nIn All Saints Chapel:")
print("1. Search for the underground passage")
print("2. Speak with the priest")
answer = input("> ")
choices.append("Chapel: " + answer)

if answer == "2":
    play_scene("4 (preist faint muttering sounds).m4a")
    print("1. Ask the priest to take the mace out of its case")
    print("2. Leave the priest alone")
    answer = input("> ")
    choices.append("Priest: " + answer)
    if answer == "1":
        print("The priest takes the mace from the case and returns it to the main room.")
        print("1. Take the mace while the priest is nearby")
        print("2. Leave it and search for the passage")
        answer = input("> ")
        choices.append("Main room: " + answer)
        if answer == "1":
            has_mace = True
            caught = True
        else:
            print("You search the chapel for the passage.")
            play_scene("2 final music for game.m4a")
    else:
        print("You leave the priest alone and search the chapel.")
        play_scene("2 final music for game.m4a")

if not caught:
    if knows_passage:
        print("You use the passage you heard about and enter the underground rooms.")
        play_scene("3 final music for game .m4a")
    elif knows_room:
        print("You know the room where the mace is stored, but not the entrance.")
        print("1. Search the chapel for the passage")
        print("2. Ask the priest where it is")
        answer = input("> ")
        choices.append("Finding the passage: " + answer)
        if answer == "1":
            print("You find the passage and enter the underground rooms.")
            play_scene("3 final music for game .m4a")
        else:
            print("The priest notices your questions and catches you.")
            caught = True
    else:
        print("You do not know where the passage is.")
        caught = True

if not caught:
    print("The mace is in a glass case in the storage room.")
    print("1. Look for the key")
    print("2. Break the glass")
    answer = input("> ")
    choices.append("Glass case: " + answer)
    if answer == "1":
        print("You find the key and open the case.")
        inventory.append("key")
        has_mace = True
    elif "rocks" in inventory or "sticks" in inventory:
        print("You break the glass with an item from the cemetery.")
        has_mace = True
        caught = True
    else:
        print("You do not have anything to break the glass with.")
        print("The priest catches you before you can take the mace.")
        caught = True

if has_mace and not caught:
    play_scene("5 stage 1 of destroying the mace.m4a")
    print("\nYou have the mace. Choose how to dispose of it.")
    print("1. Use an item to destroy it")
    print("2. Try without an item")
    answer = input("> ")
    choices.append("Mace: " + answer)
    if answer == "1" and ("shovel" in inventory or "rocks" in inventory or "sticks" in inventory):
        play_scene("6 2nd stage of destroying the mace.m4a")
        play_effect("destroying the mace.m4a")
        print("You use an item you found to destroy the mace.")
        destroyed_mace = True
        print("1. Leave quietly")
        print("2. Make enough noise to be noticed")
        answer = input("> ")
        choices.append("After destroying the mace: " + answer)
        if answer == "1":
            play_scene("8.m4a")
            print("ENDING: You destroyed the mace and got away.")
        else:
            caught = True
            play_scene("8.m4a")
            print("The priest catches you after the mace is destroyed.")
    else:
        caught = True
        print("You cannot destroy the mace. The priest catches you with it.")

if caught:
    play_scene("8.m4a")
    if destroyed_mace:
        print("ENDING: You are caught after destroying the mace.")
    elif has_mace:
        print("ENDING: You are caught with the mace before destroying it.")
    else:
        print("ENDING: You are caught before reaching the mace.")

print("\nInventory:")
if len(inventory) == 0:
    print("Nothing")
else:
    for item in inventory:
        print(item)

print("\nChoice log:")
for choice in choices:
    print(choice)
```
